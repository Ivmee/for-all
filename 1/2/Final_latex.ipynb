##The simplest code

```python
import jinja2
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, FileLink

# Create template with equations
template_content_simple = r"""
\documentclass{article}
\usepackage{graphicx}
\usepackage{booktabs}
\usepackage{amsmath}

\begin{document}
\title{<< title >>}
\author{<< author >>}
\date{\today}
\maketitle

\section{Introduction}
<< introduction >>

\section{Equations}
\begin{equation}
    e^{i\pi} + 1 = 0
\end{equation}

<< equation >>

\section{Data Analysis}
\begin{figure}[h]
    \centering
    \includegraphics[width=0.8\textwidth]{<< plot_path >>}
    \caption{Data Visualization}
\end{figure}

<< analysis_section >>

\section{Conclusion}
<< conclusion >>

\end{document}
"""

with open('report_template_simple.tex', 'w') as f:
    f.write(template_content_simple)

# Generate content
plt.figure(figsize=(8, 4))
plt.plot([1, 2, 3, 4], [1, 4, 9, 16], 'b-o')
plt.savefig('figure.png', bbox_inches='tight', dpi=300)
plt.close()

df = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar'],
    'Sales': [100, 150, 200]
})

context_simple = {
    "title": "Report with Equations",
    "author": "Colab User",
    "introduction": "This report contains both static and dynamic equations.",
    "plot_path": "figure.png",
    "equation": r"\begin{equation} \int_{0}^{1} x^2 dx = \frac{1}{3} \end{equation}",
    "analysis_section": df.to_latex(index=False),
    "conclusion": "Equations rendered successfully using LaTeX."
}

# Jinja2 environment setup
template_loader_simple = jinja2.FileSystemLoader(searchpath="./")
template_env_simple = jinja2.Environment(
    loader=template_loader_simple,
    block_start_string='{%',
    block_end_string='%}',
    variable_start_string='<<',
    variable_end_string='>>',
    comment_start_string='<#',
    comment_end_string='#>'
)

template_simple = template_env_simple.get_template("report_template_simple.tex")
output_simple = template_simple.render(context_simple)

with open("generated_report_simple.tex", "w") as f:
    f.write(output_simple)

# Compile to PDF
!pdflatex -interaction=nonstopmode generated_report_simple.tex > /dev/null 2>&1
!pdflatex -interaction=nonstopmode generated_report_simple.tex > /dev/null 2>&1

display(FileLink('generated_report_simple.pdf'))
!rm -f *.aux *.log *.out *.tex *.png
```

##Another Example


- Relative Coordinate System (0-1 scale)

- Multiple Annotation Types:

  - Text boxes with arrows

  - Reference lines

  - Measurement spans

  - Highlight circles

```python
!apt-get install texlive-latex-extra texlive-fonts-recommended texlive-science latexmk > /dev/null 2>&1
!pip install jinja2 pandas matplotlib > /dev/null

import jinja2
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, FileLink

# Create template with TikZ annotations
template_content1 = r"""
\documentclass[12pt]{article}
\usepackage{graphicx}
\usepackage{tikz}
\usepackage{booktabs}
\usepackage{amsmath}

\begin{document}
\title{\LARGE\textbf{<< title >>}}
\author{\textit{<< author >>}}
\date{\small\today}
\maketitle

\begin{figure}[h]
    \centering
    \begin{tikzpicture}
        \node[anchor=south west] (image) at (0,0) {
            \includegraphics[width=0.95\textwidth]{<< plot_path >>}
        };

        \begin{scope}[x={(image.south east)}, y={(image.north west)}]
            << tikz_annotations >>
        \end{scope}
    \end{tikzpicture}
    \caption{\small << plot_caption >>}
\end{figure}

<< analysis_section >>

\end{document}
"""

with open('report_template1.tex', 'w') as f:
    f.write(template_content1)

# Generate clean base plot
plt.figure(figsize=(10, 6))
x = [1, 2, 3, 4]
y = [1, 4, 9, 16]
plt.plot(x, y, 'b-o', linewidth=2)
plt.title('Base Data Plot', fontsize=12)
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.grid(True, alpha=0.3)
plt.savefig('figure.png', bbox_inches='tight', dpi=300)
plt.close()

# Define TikZ annotations
tikz_annotations = r"""
% Diagonal trend line
\draw[red, thick, dashed] (0.15,0.85) -- (0.85,0.15);

% Vertical reference line
\draw[blue, thick, opacity=0.7] (0.5,0) -- (0.5,1);

% Text annotation
\node[draw=black, fill=yellow, align=left] at (0.3,0.7) {
    \scriptsize Key Growth\\Phase
};
\draw[->, thick] (0.3,0.7) -- (0.4,0.6);

% Data point highlight
\draw[orange, thick] (0.75,0.25) circle [radius=0.07];

% Measurement arrow
\draw[<->, thick, purple] (0.2,0.2) -- node[midway, below] {\small Critical Range} (0.8,0.2);
"""

context1 = {
    "title": "Annotated Visualization Report",
    "author": "Data Team",
    "plot_path": "figure.png",
    "plot_caption": "Base plot with LaTeX-generated annotations",
    "tikz_annotations": tikz_annotations,
    "analysis_section": r"""
    \begin{itemize}
        \item Red dashed line shows expected trend
        \item Blue line marks midpoint
        \item Yellow box highlights key phase
        \item Orange circle indicates outlier
    \end{itemize}
    """
}

# Jinja2 setup
template_loader = jinja2.FileSystemLoader(searchpath="./")
template_env = jinja2.Environment(
    loader=template_loader,
    block_start_string='{%',
    block_end_string='%}',
    variable_start_string='<<',
    variable_end_string='>>'
)

template = template_env.get_template("report_template1.tex")
output = template.render(context1)

with open("generated_report1.tex", "w") as f:
    f.write(output)

# Compile to PDF
!pdflatex -interaction=nonstopmode generated_report1.tex > /dev/null 2>&1
!pdflatex -interaction=nonstopmode generated_report1.tex > /dev/null 2>&1

display(FileLink('generated_report1.pdf'))
!rm -f *.aux *.log *.out *.tex *.png
```
